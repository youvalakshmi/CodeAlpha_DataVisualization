## Objective

The objective of this task is to transform the Netflix dataset into meaningful visualizations using Python. Various charts and graphs are created using Matplotlib and Seaborn to identify trends, compare categories, and present insights in a clear and visually appealing manner.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
%matplotlib inline
```

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

## Visualization 1: Distribution of Content Type

### Objective

This visualization shows the distribution of Movies and TV Shows available in the Netflix dataset. It helps identify which type of content is more dominant on the platform.

```python

plt.figure(figsize=(6,5))
sns.countplot(data=df, x="type", hue="type", palette="Set2", legend=False)

plt.title("Distribution of Netflix Content Type")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

### Observation

- Movies are available in much higher numbers than TV Shows.
- Netflix's content library is dominated by movies.
- This indicates that the platform focuses more on movie-based entertainment than television series.

## Visualization 2: Top 10 Countries with the Most Netflix Content

### Objective

This visualization identifies the top 10 countries that have contributed the highest number of Movies and TV Shows to Netflix. It helps understand the geographical distribution of Netflix content.

```python
# Top 10 countries with the most content

top_countries = (
    df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    hue=top_countries.index,
    palette="viridis",
    legend=False
)

plt.title("Top 10 Countries with the Most Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

### Observation

- The United States contributes the highest number of titles available on Netflix.
- India is also one of the leading content-producing countries.
- This shows that Netflix has a strong presence in countries with large entertainment industries.

## Visualization 3: Netflix Content Released Over the Years

### Objective

This visualization shows how the number of Netflix titles has changed over the years. It helps identify growth trends and understand how Netflix's content library has expanded over time.

```python

release_year = df["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    x=release_year.index,
    y=release_year.values,
    marker="o",
    color="blue"
)

plt.title("Netflix Content Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

### Observation

- The number of Netflix titles increased significantly after 2010.
- Content production reached its highest levels during the late 2010s.
- This trend indicates Netflix's rapid expansion and growing investment in entertainment content.

## Visualization 4: Top 10 Genres on Netflix

### Objective

This visualization displays the top 10 most common genres available on Netflix. It helps understand the types of content that dominate the platform and the viewing preferences of its audience.

```python

top_genres = (
    df["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    hue=top_genres.index,
    palette="magma",
    legend=False
)

plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

### Observation

- International Movies and Dramas are among the most common genres on Netflix.
- The platform offers a wide variety of genres to suit different audience preferences.
- This indicates that Netflix focuses on providing diverse entertainment content for viewers worldwide.

## Visualization 5: Distribution of Content Ratings

### Objective

This visualization shows the distribution of content ratings in the Netflix dataset. It helps understand the target audience for Netflix content based on rating categories.

```python

rating_counts = df["rating"].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values,
    hue=rating_counts.index,
    palette="coolwarm",
    legend=False
)

plt.title("Distribution of Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

### Observation

- TV-MA and TV-14 are the most common content ratings on Netflix.
- This indicates that a large portion of Netflix content is intended for mature and teenage audiences.
- Fewer titles are available for younger children compared to adult viewers.

## Visualization 6: Content Added to Netflix Over the Years

### Objective

This visualization shows how many Movies and TV Shows were added to Netflix each year. It helps analyze the growth of Netflix's content library over time.

```python
# Convert date_added to datetime
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

# Extract year
added_year = df["date_added"].dt.year.value_counts().sort_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    x=added_year.index,
    y=added_year.values,
    marker="o",
    color="green"
)

plt.title("Content Added to Netflix Over the Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")

plt.xticks(rotation=45)

plt.show()
```

### Observation

- The number of titles added to Netflix increased rapidly after 2015.
- The platform expanded its content library significantly during the late 2010s.
- This reflects Netflix's growing investment in providing a wide range of entertainment content.

## Visualization 7: Distribution of Movie Duration

### Objective

This visualization shows the distribution of movie durations available on Netflix. It helps understand the common length of movies and identify whether most movies are short, medium, or long in duration.

```python
# Select only movies
movies = df[df["type"] == "Movie"].copy()

# Extract duration in minutes
movies["duration"] = movies["duration"].str.replace(" min", "", regex=False)
movies["duration"] = pd.to_numeric(movies["duration"], errors="coerce")

plt.figure(figsize=(10,6))

sns.histplot(
    movies["duration"].dropna(),
    bins=30,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Movie Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

### Observation

- Most Netflix movies have a duration between 80 and 120 minutes.
- Very short and very long movies are less common.
- This indicates that feature-length movies make up the majority of Netflix's movie collection.

## Visualization 8: Missing Values Heatmap

### Objective

This visualization highlights the missing values present in the Netflix dataset. It helps identify columns with incomplete data that may require cleaning before performing advanced analysis or building machine learning models.

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")

plt.show()
```

### Observation

- Some columns such as **director**, **cast**, **country**, and **date_added** contain missing values.
- Most other columns have complete or nearly complete information.
- Identifying missing values is an important step before performing advanced data analysis or predictive modeling.

## Visualization 9: Netflix Dashboard Summary

### Objective

This dashboard combines multiple visualizations into a single figure to provide a quick overview of the Netflix dataset. It helps summarize key information and enables better decision-making through visual analysis.

```python
plt.figure(figsize=(16,10))

# 1. Content Type
plt.subplot(2,2,1)
sns.countplot(data=df, x="type", hue="type", palette="Set2", legend=False)
plt.title("Content Type")

# 2. Ratings
plt.subplot(2,2,2)
rating_counts = df["rating"].value_counts().head(5)
sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values,
    hue=rating_counts.index,
    palette="coolwarm",
    legend=False
)
plt.xticks(rotation=45)
plt.title("Top Content Ratings")

# 3. Release Trend
plt.subplot(2,2,3)
release_year = df["release_year"].value_counts().sort_index()
plt.plot(release_year.index, release_year.values, marker="o")
plt.title("Release Trend")
plt.xlabel("Year")

# 4. Movie Duration
plt.subplot(2,2,4)
sns.histplot(
    movies["duration"].dropna(),
    bins=20,
    color="skyblue",
    edgecolor="black"
)
plt.title("Movie Duration")

plt.tight_layout()
plt.show()
```

### Observation

- The dashboard provides a quick overview of the Netflix dataset.
- Movies are more common than TV Shows.
- Most content belongs to TV-MA and TV-14 ratings.
- Netflix expanded rapidly in recent years.
- Most movies have durations between 80 and 120 minutes.
